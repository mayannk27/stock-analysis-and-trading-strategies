```python
# -*- coding: utf-8 -*-

"""
Created on Tue Sep 24 12:02:16 2019
@author: Yizhen Zhao
Kernel Density
"""
import numpy as np
from scipy.optimize import minimize
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
from datetime import datetime


import yfinance as yf

start_date = datetime(2021,1,1) # 2019 1 1
end_date = datetime(2021,12,31)

plt.figure(1)
STOCK = yf.download('TSLA',start_date ,end_date)
# X2 = yf.download('AAPL',start_date ,end_date)
# corr = np.corrcoef(X1['Open'].values,X2['Open'].values)
# print('correlation between the two', corr)

sns.kdeplot(data=STOCK['Adj Close'].values)
STOCK['Adj Close'].describe()

plt.figure(2)
MONTH = STOCK.resample('M').mean()
sns.kdeplot(data=MONTH['Adj Close'])
MONTH['Adj Close'].describe()
```
